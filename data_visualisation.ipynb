```python
# Import pandas
import pandas as pd
```

```python
# Define the path where data is kept
path = '/content/drive/MyDrive/Copy of pp-2018.csv'
```

```python
df = pd.read_csv(path)
df.head()
```

```python
df = pd.read_csv(path, names = ["TID","Price","Sale Date", "Postcode",
                                    "Property Type","New Build","Tenure",
                                    "PAON","SAON","Street","Locality",
                                    "Town/City","District","County",
                                    "PPD Category","Record Status"],
                 parse_dates= ["Sale Date"])
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df['month'] = df['Sale Date'].dt.month
df['day'] = df['Sale Date'].dt.day
df.head()
```

```python
df['month'].value_counts()
```

```python
month_sale = df['month'].value_counts().sort_index()
month_sale
```

```python
month_sale.plot(kind = 'bar')
```

```python
from matplotlib import pyplot as plt
```

```python
plt.figure(figsize = (9,3))
month_sale.plot(kind = 'bar')
```

```python
import seaborn as sns
```

```python
plt.figure(figsize = (9,3))
month_sale.plot(kind = 'bar',color = sns.color_palette('Set2'))
```

```python
plt.figure(figsize = (9,3))
month_sale.plot(kind = 'line',color = sns.color_palette('Set2'))
plt.title('Monthly Sales Volume')
plt.xlabel('Month')
plt.ylabel('Sales Volume')
```

```python
plt.figure(figsize = (9,3))
month_sale.plot(kind = 'bar',color = sns.color_palette('Set2'))
plt.title('Monthly Sales Volume')
plt.xlabel('Month')
plt.ylabel('Sales Volume')
```

```python
plt.figure(figsize = (9,3))
month_sale.plot(kind = 'pie',color = sns.color_palette('Set2'))
plt.title('Monthly Sales Volume')
plt.xlabel('Month')
plt.ylabel('Sales Volume')
```
