```python
def reversewords(s):

   words = s.split()
   words = words[::-1]

   result = " ".join(words)

   return result
s = "hello world python"
print(reversewords(s))
```

two way to remove space from string
# without using replace()
# with replace()

```python
spell = "i love dancing"

result = ""

for i in spell:
    if i != " ":
      result = result + i

print(result)
```

```python
print(spell.replace(" ",""))
```

```python
spell = "i hate dancing"
print(spell.replace("hate","love"))
```

#code without using replace() using logic

```python
spell = "i like iphone"
word = spell.split()


result = ""

for i in word:
    if i  == "iphone":
     result = result + "samsung"
    else:
      result = result + i +  " "

print(result)
```
